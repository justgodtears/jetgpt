```python
import pandas as pd
from tokenizers import Tokenizer

tokenizer = Tokenizer.from_file("../data/tokenized/tokenizer_24k.json")
df = pd.read_parquet("../data/processed/processed_data__09-00__09_06_2026.parquet")
sample = df["text"].sample(10000, random_state=42)
lengths = [len(tokenizer.encode(text).ids) for text in sample]

unk_id = tokenizer.token_to_id("[UNK]")
unk_list = []
for text in sample:
    encoded = tokenizer.encode(text)
    unk_list.append(encoded.ids.count(unk_id))

percentage_unk = sum(unk_list) / sum(lengths)

print(pd.Series(lengths).describe())
print(percentage_unk)
```
